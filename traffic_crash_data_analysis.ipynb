Query-1 : Find the top 5 most dangerous combinations of weather and crash type based on total crashes.

```python
from tabulate import tabulate
cursor.execute('''SELECT WEATHER_CONDITION,
                          CRASH_TYPE, 
                          COUNT(*) AS total_crashes
                   FROM traffic_crash_data
                   WHERE INJURIES_TOTAL > 0
                   GROUP BY WEATHER_CONDITION, 
                            CRASH_TYPE
                   ORDER BY total_crashes DESC
                  LIMIT 5;''')
result1 = cursor.fetchall()

# Define table headers
headers = ["WEATHER_CONDITION", "CRASH_TYPE","TOTAL_CRASHES"]

# Print the result as a table
print(tabulate(result1, headers=headers, tablefmt="grid"))

connection.commit()
```

Query-2 : Identify the top 10 streets with the highest number of injury crashes

```python
cursor.execute('''SELECT STREET_NO,
                       STREET_DIRECTION,
                       STREET_NAME,
                       CRASH_TYPE, 
                       COUNT(INJURIES_TOTAL) as total_injuries 
                FROM traffic_crash_data 
                WHERE INJURIES_TOTAL > 0 
                GROUP BY STREET_NO,
                         STREET_DIRECTION,
                         STREET_NAME,
                         CRASH_TYPE
                ORDER BY COUNT(INJURIES_TOTAL) DESC 
                LIMIT 10;''')

result2 = cursor.fetchall()

# Define table headers
headers = ["STREET_NO", "STREET_DIRECTION","STREET_NAME","CRASH_TYPE","TOTAL_INJURIES"]

# Print the result as a table
print(tabulate(result2, headers=headers, tablefmt="grid"))


connection.commit()
```

Query-3 : Find the percentage of crashes that resulted in injuries for each crash type

```python
cursor.execute(''' 
                SELECT CRASH_TYPE, 
                       COUNT(*) AS total_crashes, 
	                   SUM(CASE WHEN INJURIES_TOTAL > 0 THEN 1 ELSE 0 END) AS injury_crashes,
	                   ROUND(100.0 * SUM(CASE WHEN INJURIES_TOTAL > 0 THEN 1 ELSE 0 END) / COUNT(*),2) AS injury_crashes_percentage
                FROM traffic_crash_data
                GROUP BY CRASH_TYPE
                ORDER BY injury_crashes_percentage DESC;
            ''')

result3 = cursor.fetchall()

# Define table headers
headers = ["CRASH_TYPE", "TOTAL_CRASHES","INJURY_CRASHES","INJURY_CRASHES_PERCENTAGE"]

# Print the result as a table
print(tabulate(result3, headers=headers, tablefmt="grid"))

connection.commit()
```

The analysis shows which crash types are most likely to cause injuries, helping prioritize safety measures for the most dangerous accident types.

Query-4 : Determine the peak crash hour for each month

```python
cursor.execute(''' WITH monthly_crash_hours AS
                (SELECT CRASH_MONTH,
                        CRASH_HOUR,
                        COUNT(*) AS crash_count,
                 ROW_NUMBER() OVER(PARTITION BY CRASH_MONTH ORDER BY COUNT(*) DESC) AS rn
                 FROM traffic_crash_data 
                 GROUP BY CRASH_MONTH,CRASH_HOUR
              )

              SELECT CRASH_MONTH, 
                     CRASH_HOUR AS peak_hour, 
                     crash_count
              FROM monthly_crash_hours
              WHERE rn=1
              ORDER BY CRASH_MONTH;''')

result4 = cursor.fetchall()

# Define table headers
headers = ["CRASH_MONTH", "PEAK_HOUR","CRASH_COUNT"]

# Print the result as a table
print(tabulate(result4, headers=headers, tablefmt="grid"))

connection.commit()
```

The analysis identifies the busiest crash hours for each month, helping understand when accidents are most likely to happen throughout the year.
These peak crash periods can help traffic authorities plan better road safety measures.

Query-9 : Find the top 5 streets with the highest injury rate, considering only streets with more than 100 crashes

```python
cursor.execute('''SELECT STREET_NAME, 
                       COUNT(*) AS total_crashes, 
                       SUM(CASE WHEN INJURIES_TOTAL > 0 THEN 1 ELSE 0 END) AS injury_crashes,
                       ROUND(100.0 * SUM(CASE WHEN INJURIES_TOTAL > 0 THEN 1 ELSE 0 END) / COUNT(*),2) AS injury_rate_percentage
             FROM traffic_crash_data
             GROUP BY STREET_NAME
             HAVING COUNT(*) > 100
             ORDER BY injury_rate_percentage DESC
             LIMIT 5;''')

result9 = cursor.fetchall()

# Define table headers
headers = ["STREET_NAME", "TOTAL_CRASHES","INJURY_CRASHES","INJURY_RATE_PERCENTAGE"]

# Print the result as a table
print(tabulate(result9, headers=headers, tablefmt="grid"))

connection.commit()
```

Query-10 : For each year, identify the most common crash type. 

```python
cursor.execute('''WITH crash_counts AS ( 
                SELECT YEAR, 
                       CRASH_TYPE, 
                       COUNT(*) AS total_crashes,
	                RANK() OVER (PARTITION BY YEAR ORDER BY COUNT(*) DESC) AS rank_num
              FROM traffic_crash_data
              GROUP BY YEAR,CRASH_TYPE
       )
       SELECT YEAR, 
              CRASH_TYPE,
              total_crashes
       FROM crash_counts
       WHERE rank_num = 1
       ORDER BY YEAR DESC;''')

result10 = cursor.fetchall()

# Define table headers
headers = ["YEAR", "CRASH_TYPE","TOTAL_CRASHES"]

# Print the result as a table
print(tabulate(result10, headers=headers, tablefmt="grid"))

connection.commit()
```

The analysis shows the most common crash type each year, helping identify recurring accident patterns and areas where road safety efforts should be focused.

Query-11 : Find the day of the week with the highest average crashes per hour

```python
cursor.execute('''WITH hourly_crashes AS(
                SELECT CRASH_DAY_OF_WEEK,
                       CRASH_HOUR,
                       COUNT(*) AS crashes_per_hour,
                       ROW_NUMBER() OVER (PARTITION BY CRASH_DAY_OF_WEEK ORDER BY COUNT(*) DESC) AS rn
                FROM traffic_crash_data 
                GROUP BY CRASH_DAY_OF_WEEK,
                         CRASH_HOUR
              )

              SELECT CRASH_DAY_OF_WEEK,
                     CRASH_HOUR,
                     ROUND(AVG(crashes_per_hour), 0) AS avg_crashes_per_hour 
              FROM hourly_crashes 
              WHERE rn=1
              GROUP BY CRASH_DAY_OF_WEEK,
                       CRASH_HOUR
              ORDER BY avg_crashes_per_hour DESC;''')

result11 = cursor.fetchall()

# Define table headers
headers = ["CRASH_DAY_OF_WEEK", "CRASH_HOUR","AVG_CRASHES_PER_HOUR"]

# Print the result as a table
print(tabulate(result11, headers=headers, tablefmt="grid"))

connection.commit()
```

The analysis shows which day of the week has the highest crash activity per hour, helping identify the most accident-prone days for better traffic control and safety planning.

Query-12 : Identify high-risk time slots:
Group hours into buckets (Morning, Afternoon, Evening, Night)
Find which bucket has the highest injury crashes


```python
cursor.execute('''
                SELECT
                  time_bucket,
                  COUNT(*) AS injury_crashes
                FROM (SELECT
                        CASE
                            WHEN CRASH_HOUR BETWEEN 6 AND 11 THEN 'Morning'
                            WHEN CRASH_HOUR BETWEEN 12 AND 17 THEN 'Afternoon'
                            WHEN CRASH_HOUR BETWEEN 18 AND 23 THEN 'Evening'
                            ELSE 'Night'
                        END AS time_bucket
                FROM traffic_crash_data
                WHERE INJURIES_TOTAL > 0
            )AS grouped_times
            GROUP BY time_bucket
            ORDER BY injury_crashes DESC;''')

result12 = cursor.fetchall()

# Define table headers
headers = ["TIME_BUCKET", "INJURY_CRASHES"]

# Print the result as a table
print(tabulate(result12, headers=headers, tablefmt="grid"))

connection.commit()
```

The analysis shows which time of day has the most injury crashes, helping authorities focus safety measures during the most risky hours.

Query-13 : Find the top 3 contributing causes for each crash type.
(Use window functions like ROW_NUMBER() or RANK())

```python
cursor.execute('''
               WITH contributing_cause AS ( 
               SELECT 
                    PRIM_CONTRIBUTORY_CAUSE, 
                    SEC_CONTRIBUTORY_CAUSE, 
                    CRASH_TYPE, 
                    COUNT(*) AS total_crashes,
	                ROW_NUMBER() OVER (PARTITION BY CRASH_TYPE ORDER BY COUNT(*) DESC) AS rn
                FROM traffic_crash_data
                GROUP BY 
                      PRIM_CONTRIBUTORY_CAUSE,
                      SEC_CONTRIBUTORY_CAUSE,
                      CRASH_TYPE
            )
            SELECT 
                  PRIM_CONTRIBUTORY_CAUSE, 
                  SEC_CONTRIBUTORY_CAUSE, 
                  CRASH_TYPE, 
                  total_crashes
            FROM contributing_cause
            WHERE rn <= 3
            ORDER BY total_crashes DESC
            LIMIT 3;''')

result13 = cursor.fetchall()

# Define table headers
headers = ["PRIM_CONTRIBUTORY_CAUSE", "SEC_CONTRIBUTORY_CAUSE","CRASH_TYPE","TOTAL_CRASHES"]

# Print the result as a table
print(tabulate(result13, headers=headers, tablefmt="grid"))

connection.commit()
```

The analysis shows the main causes behind each type of crash, helping identify the most common reasons accidents happen so safety measures can be improved.

Query-14 : Calculate the year-over-year growth rate of crashes.
(Use LAG() window function)

```python
cursor.execute('''
               WITH yearly_crashes AS (
               SELECT
                    YEAR,
                    COUNT(*) AS total_crashes
               FROM traffic_crash_data
               GROUP BY YEAR
        ),
        previous_year_data AS (
            SELECT 
                YEAR,
                total_crashes,
                LAG(total_crashes) OVER (ORDER BY YEAR) AS previous_year_crashes
            FROM yearly_crashes
        )
        SELECT
               YEAR,
               total_crashes,
               previous_year_crashes,
               ROUND((
                       (total_crashes - previous_year_crashes)
                        * 100.0) / previous_year_crashes,2) AS yoy_growth_rate_percentage
               FROM previous_year_data
               ORDER By YEAR;
''')

result14 = cursor.fetchall()

# Define table headers
headers = ["YEAR", "TOTAL_CRASHES","PREVIOUS_YEAR_CRASHES","YOY_GROWTH_RATE_PERCENTAGE"]

# Print the result as a table
print(tabulate(result14, headers=headers, tablefmt="grid"))

connection.commit()
```

The analysis shows whether crash numbers are increasing or decreasing each year, helping track road safety trends over time.

Query-15 : Identify hotspot zones:
Group nearby locations (round latitude & longitude to 2 decimal places)
Find top 10 zones with highest crashes

```python
cursor.execute('''
                SELECT 
                    ROUND(LATITUDE, 2) AS latitude_zone,
                    ROUND(LONGITUDE, 2) AS longitude_zone,
                    COUNT(*) AS total_crashes
                FROM traffic_crash_data
                GROUP BY 
                    ROUND(LATITUDE, 2),
                    ROUND(LONGITUDE, 2)
                ORDER BY total_crashes DESC
                LIMIT 10;
        ''')

result15 = cursor.fetchall()

# Define table headers
headers = ["LATITUDE", "LONGITUDE","TOTAL_CRASHES"]

# Print the result as a table
print(tabulate(result15, headers=headers, tablefmt="grid"))

connection.commit()
```

The analysis shows the top 10 crash hotspot zones where accidents happen most often, helping authorities focus on improving road safety in those specific areas.
